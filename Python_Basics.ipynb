```python
a = 10
b = 3
print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.333... (float division)
print(a // b)  # 3 (integer division)
print(a % b)   # 1 (remainder)
```

```python
marks = 75

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Grade: {grade}")
```

```python
# for loop
for i in range(5):
    print(i)   # 0, 1, 2, 3, 4

# while loop
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
fruits = ["apple", "banana", "mango"]
print(fruits[0])        # apple
fruits.append("orange") # add karo
print(len(fruits))      # 4 — length

for fruit in fruits:
    print(fruit)
```

```python
name = "Ahmed"          # string
age = 21                 # integer
gpa = 3.4                # float
is_student = True        # boolean

print(name, age, gpa, is_student)
print(type(age))
```

```python
for number in range(1, 21):
    if number % 2 == 0:
        print(number, "(even)")
    else:
        print(number)
```

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Ahmed"))   # Hello, Ahmed!
print(greet("Sara"))    # Hello, Sara!
```

```python
def add_numbers(a, b=10):
    return a + b

print(add_numbers(5, 3))
print(add_numbers(5))
```

```python
def is_even(number):
    if number % 2 == 0:
        return True
    else:
        return False

print(is_even(4))   # True
print(is_even(7))   # False
```

```python
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")
```

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
squares = []
for x in range(1, 6):
    squares.append(x * x)
print(squares)   # [1, 4, 9, 16, 25]
```

```python
squares = [x * x for x in range(1, 6)]
print(squares)   # [1, 4, 9, 16, 25]
```

```python
even_numbers = [x for x in range(1, 21) if x % 2 == 0]
print(even_numbers)   # [2, 4, 6, ..., 20]
```

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print(arr)
print(type(arr))   # numpy.ndarray

# 2D array (matrix jaisa)
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix)
arr = np.array([1, 2, 3, 4, 5])

print(arr + 10)      # [11 12 13 14 15]
print(arr * 2)       # [2 4 6 8 10]
print(arr > 3)        # [False False False True True]
arr = np.array([4, 8, 2, 10, 6])

print(arr.sum())     # 30 - total
print(arr.mean())    # 6.0 - average
print(arr.max())     # 10
print(arr.min())     # 2
print(arr.shape)     # (5,)
print(len(arr))      # 5
arr = np.array([10, 20, 30, 40, 50])

print(arr[0])      # 10 (pehla element)
print(arr[-1])     # 50 (last element)
print(arr[1:4])    # [20 30 40]
arr = np.array([1, 2, 3, 4, 5, 6])
reshaped = arr.reshape(2, 3)   # 2 rows, 3 columns
print(reshaped)
# [[1 2 3]
#  [4 5 6]]
```
